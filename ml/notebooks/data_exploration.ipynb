# Data Exploration

```python
# ================================
# IMPORTS
# ================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
df = pd.read_csv("../data/processed/building_level.csv")
print(f"Dataset: {df.shape[0]} rows x {df.shape[1]} columns")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
print(df["accessibility_class"].value_counts())

sns.countplot(x="accessibility_class", data=df)
plt.title("Accessibility Class Distribution")
plt.show()
```

```python
sns.histplot(df["total_floors"], bins=10, kde=True)
plt.title("Distribution of Total Floors")
plt.show()
```

```python
sns.countplot(x="elevator_present", hue="accessibility_class", data=df)
plt.title("Elevator Presence vs Accessibility Class")
plt.show()
```

```python
sns.countplot(x="step_free_entrance", hue="accessibility_class", data=df)
plt.title("Step-Free Entrance vs Accessibility Class")
plt.show()
```

```python
sns.histplot(df["total_services"], bins=20, kde=True)
plt.title("Total Services Distribution")
plt.show()
```

```python
sns.histplot(df["num_services_upper"], bins=20, kde=True)
plt.title("Services Located on Upper Floors")
plt.show()
```

```python
sns.histplot(df["num_critical_upper"], bins=20, kde=True)
plt.title("Critical Services on Upper Floors")
plt.show()
```

```python
sns.boxplot(x="elevator_present", y="num_services_upper", data=df)
plt.title("Upper Floor Services vs Elevator Presence")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(df["avg_clear_signage"], bins=10, kde=True, ax=axes[0])
axes[0].set_title("Avg Clear Signage Distribution")

sns.histplot(df["avg_high_contrast"], bins=10, kde=True, ax=axes[1])
axes[1].set_title("Avg High Contrast Signage Distribution")

plt.tight_layout()
plt.show()
```

```python
ratio_cols = [
    "accessible_service_ratio",
    "upper_service_ratio",
    "critical_upper_ratio",
    "accessible_critical_ratio"
]

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
for ax, col in zip(axes.flatten(), ratio_cols):
    sns.histplot(df[col], bins=15, kde=True, ax=ax)
    ax.set_title(f"Distribution of {col}")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
for ax, col in zip(axes.flatten(), ratio_cols):
    sns.boxplot(x="accessibility_class", y=col, data=df,
                order=["low", "medium", "high"], ax=ax)
    ax.set_title(f"{col} by Accessibility Class")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(x="accessibility_class", y="avg_clear_signage", data=df,
            order=["low", "medium", "high"], ax=axes[0])
axes[0].set_title("Avg Clear Signage by Class")

sns.boxplot(x="accessibility_class", y="avg_high_contrast", data=df,
            order=["low", "medium", "high"], ax=axes[1])
axes[1].set_title("Avg High Contrast by Class")

plt.tight_layout()
plt.show()
```

```python
sns.boxplot(x="accessibility_class", y="num_critical_upper", data=df)
plt.title("Critical Upper Services vs Accessibility Class")
plt.show()
```

```python
# Drop non-numeric columns for correlation
numeric_df = df.drop(columns=["accessibility_class", "site_type"])

plt.figure(figsize=(14, 10))
sns.heatmap(numeric_df.corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# Site type distribution (now a string column)
sns.countplot(x="site_type", data=df, order=df["site_type"].value_counts().index)
plt.title("Distribution of Site Types")
plt.show()
```
